```python
# criando uma lista com todas as tabelas que nos foi fornecida:
def filter_file(directorio, extensao):
    list_direct = os.listdir(directorio)
    list_filter = list(filter(lambda filer: filer.endswith(extensao), list_direct))
    return list_filter
    
```
